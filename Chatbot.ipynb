### <span style="color:black"> Conditional Statements </span>

- Conditional statements controls flow of the conversation between the user and the chatbotby responding differently based on the user input.
- if, elif, and else are used to match user input such that chatbot can give a suitable response to the users input.For example, if the user expresses that they are feeling "sad",the chatbot will respond with words of encouragement like "Writing about it may help. Would you like to share?"
- The chatbot will have a suitable response for given list of feelings
- This provides them users with a form of mental and emotional support


```python
# Function to get user input and respond
def chatbot():
    
    print("🧠 Welcome to WellnessBot")
    print("I'm here to support your mental and emotional well-being.")
    print("Type 'exit' anytime to end the conversation.\n")
    
    # Variable to store user's name
    user_name = input("WellnessBot: What should I call you? ")

    print("WellnessBot: Hello " + user_name + ". How are you feeling today?")
    print("You can say: happy, sad, stressed, anxious, nervous, scared, tired")
    print("Or type 'write' to write in your diary, or 'view' to see past entries.\n")

    # Main loop
    while True:
        user_input = input("You: ").lower().strip()

        if user_input == "exit":
            print("WellnessBot: Thank you for sharing today, " + user_name + ". Take care 💙")
            break

        elif user_input in precepts:
            index = precepts.index(user_input)

            print("WellnessBot:", responses[index][0].format(user_name))
            print("WellnessBot:", responses[index][1])

            if user_input != "happy":
                entry = input("Diary entry: ")
                diary_entries.append(entry)
                print("WellnessBot: Thank you for sharing. Remember, tough moments don't last forever.")

        elif user_input == "write":
            print("WellnessBot: Go ahead and write what's on your mind.")
            entry = input("Diary entry: ")
            diary_entries.append(entry)
            print("WellnessBot: Thank you for sharing. Your feelings matter.")

        elif user_input == "view":
            if len(diary_entries) == 0:
                print("WellnessBot: Your diary is empty.")
            else:
                print("📔 Your diary entries:")
                for entry in diary_entries:
                    print("- " + entry)

        else:
            print("WellnessBot: Thank you for sharing, " + user_name + ". I'm here to listen.")
```

### <span style="color:black"> Mapping inputs to responses using dictionaries</span>

I will now add on to my chatbot by using dictionaries to map user intents to theitr corresponding responses. Instead of writing a new **dictionary** use the **ZIP** function on a list. 

```python
# Function to map moods to responses using zip
def zip_responses(precepts, responses):
    """
    Maps each mood (precept) to its corresponding response(s)
    and returns a dictionary.
    """
    mapping = {}
    for p, r in zip(precepts, responses):
        mapping[p] = r
    return mapping

# Create the dictionary
mood_map = zip_responses(precepts, responses)

```

### <span style="color:black"> Pattern-Matching using Regular Expression</span>

- Match complete words using regular expressions. 
- Example, match all occurrences of hello, hi, hey etc under a pattern called greetings.
- Use "|" symbol to string together multiple words.
- Each pattern should have atleast or 5 different occurrences of the word you want to match.
- Create atleast 5 different patterns.
  

```python
# Pattern Matching using Regular Expressions

import re

# 1️⃣ Define regex patterns (at least 5 patterns, each with ≥5 words)

patterns = {
    "greetings": r"\b(hi|hello|hey|hola|howdy)\b",
    "happy": r"\b(happy|great|good|fantastic|awesome)\b",
    "sad": r"\b(sad|down|unhappy|depressed|low)\b",
    "stressed": r"\b(stressed|pressure|overwhelmed|burnt|tense)\b",
    "anxious": r"\b(anxious|worried|nervous|uneasy|panic)\b",
    "tired": r"\b(tired|exhausted|sleepy|fatigued|drained)\b"
}
```

```python
def match_pattern(user_input):
    """
    Matches user input against predefined regex patterns
    and returns the matched category.
    """
    user_input = user_input.lower()

    for category, pattern in patterns.items():
        if re.search(pattern, user_input):
            return category

    return None
```

#### <span style="color:black"> Reusable Response Function</span>

- Dictionary mapping (mood_map)
- String formatting
- Reusable function logic

```python
def respond_to_mood(user_name, mood):
    """
    Responds using mood_map and stores diary entries if needed.
    """
    if mood in mood_map:
        for response in mood_map[mood]:
            print(response.format(user_name) if '{}' in response else response)

        # Ask for diary entry except greetings/happy
        if mood not in ["hi", "happy"]:
            entry = input("Diary entry: ").strip()
            if entry:
                diary_entries.append(entry)
                print("WellnessBot: Thank you for sharing. Your feelings matter.")
```

```python
# Function to get user input and respond
def chatbot():
    
    print("🧠 Welcome to WellnessBot")
    print("I'm here to support your mental and emotional well-being.")
    print("Type 'exit' anytime to end the conversation.\n")
    
    # Variable to store user's name
    user_name = input("WellnessBot: What should I call you? ")

    print("WellnessBot: Hello " + user_name + ". How are you feeling today?")
    print("You can say: happy, sad, stressed, anxious, nervous, scared, tired")
    print("Or type 'write' to write in your diary, or 'view' to see past entries.\n")

    while True:
        user_input = input("You: ").strip()

        # Exit condition
        if user_input.lower() == "exit":
            print(f"WellnessBot: Thank you for sharing today, {user_name}. Take care 💙")
            break

        # View diary entries
        elif user_input.lower() == "view":
            if not diary_entries:
                print("WellnessBot: Your diary is empty.")
            else:
                print("📔 Your diary entries:")
                for entry in diary_entries:
                    print("- " + entry)

        # 🔍 Detect mood using regex (Part 2.2)
        else:
            detected_mood = match_pattern(user_input)

            # 💬 Respond using dictionary mapping (Part 2.1)
            if detected_mood:
                respond_to_mood(user_name, detected_mood)
            else:
                print(f"WellnessBot: Thank you for sharing, {user_name}. I'm here to listen.")
```

### <span style="color:black"> Cleaning and Tokenizing User Input (Punctuation and Stopwords)</span>


In this part, I enhanced the chatbot by cleaning and preprocessing user input.
This includes:

- Converting user input to lowercase (case-insensitive matching)
- Removing punctuation marks
- Removing common stopwords such as *is*, *the*, *and*, *to*


```python
# Initialize stemmer and lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Preprocess user input: lowercase, remove punctuation, remove stopwords, stem, lemmatize
def preprocess_input(text):
    # Lowercase
    text = text.lower()
    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))
    # Tokenize and remove stopwords
    tokens = [word for word in text.split() if word not in stop_words]
    # Apply stemming and lemmatization
    stemmed = [stemmer.stem(word) for word in tokens]
    lemmatized = [lemmatizer.lemmatize(word) for word in tokens]
    return {
        "original": text,
        "tokens": tokens,
        "stemmed": stemmed,
        "lemmatized": lemmatized
    }

# Get random response for a given tag
def get_response(tag):
    for intent in intents_data["intents"]:
        if intent["tag"] == tag:
            return random.choice(intent["responses"])
    return None

# Find intent keyword in preprocessed input
def find_intent(user_input):
    user_input_proc = preprocess_input(user_input)
    words_to_check = user_input_proc["stemmed"] + user_input_proc["lemmatized"] + user_input_proc["tokens"]
    
    for intent in intents_data["intents"]:
        tag = intent["tag"]
        if tag in words_to_check:
            return tag
    return None
```

```python
def chatbot():
    print("🧠 Welcome to WellnessBot")
    print("I'm here to support your mental and emotional well-being.")
    print("Type 'exit' anytime to end the conversation.\n")
    
    user_name = input("WellnessBot: What should I call you? ").strip()
    print(f"WellnessBot: Hello {user_name}. How are you feeling today?")
    print("You can say: happy, sad, stressed, anxious, nervous, scared, tired")
    print("Or type 'write' to write in your diary, or 'view' to see past entries.\n")
    
    
    while True:
        user_input = input("You: ").strip()

        if user_input.lower() == "exit":
            print(f"WellnessBot: Take care, {user_name} 💙")
            break

        elif user_input.lower() == "view":
            if not diary_entries:
                print("WellnessBot: Your diary is empty.")
            else:
                print("📔 Your diary entries:")
                for entry in diary_entries:
                    print("- " + entry)

        elif user_input.lower() == "write":
            entry = input("WellnessBot: Write what's on your mind: ").strip()
            if entry:
                diary_entries.append(entry)
                save_diary()
                print("WellnessBot: Thanks for sharing 💙")

        else:
            # Preprocess input
            processed = preprocess_input(user_input)
            # Detect intent using stemmed/lemmatized/tokenized words
            intent = find_intent(user_input)

            if intent:
                response = get_response(intent)
                if response:
                    print(f"WellnessBot: {response.format(user_name)}")

                # Emotional intents trigger diary + joke + quote
                emotional_intents = ["sad", "stressed", "anxious", "tired", "upset", "hopeless"]
                if intent in emotional_intents:
                    entry = input("Diary entry: ").strip()
                    if entry:
                        diary_entries.append(entry)
                        save_diary()
                        print("WellnessBot: Thanks for sharing 💙")

                    # Tell joke
                    joke = get_response("joke")
                    if joke:
                        print(f"WellnessBot (Joke): {joke}")

                    # Show quote
                    quote = get_response("quote")
                    if quote:
                        print(f"WellnessBot (Quote): {quote}")
            else:
                print(f"WellnessBot: Thank you for sharing, {user_name}. I'm here to listen.")
   
```
